```python
"""
Ice Displacement Analysis 
Author: Hannah Picton 
Revised: 9th November 2026
Description: This script calculates and prints statistics regarding changes in annual ice displacement, and mean winter ice velocity, at high-elevation 
sites along the K-Transect. Seasonal and annual ice displacement data for LEV5 and LEV6 between 2009 and 2013 (recalculated from Tedstone & Nienow, 2018; https://doi.org/10.5285/1f69fba3-4c62-47ad-8119-08cfeec05e46, with summer defined as 1 May - 31 August) are provided in data/seasonal_annual_2009_2013/. KAN_U data for 2009-2012 are from Doyle et al. (2014; https://doi.org/10.1002/2013GL058933)  and should be added to this folder. Data for 2021-2023 can be downloaded from Tedstone et al. (2026; https://doi.org/10.5281/zenodo.18611668) and should be placed in data/seasonal_annual_2020_2024/. Ice displacement data for F003 and F004 in 1996 can be found at https://ubwp.buffalo.edu/glyremotesensinglab/2024/05/22/mass-balance-and-velocity-data-for-greenlands-ice-sheets-at-high-elevations-data/ and are entered directly below.
"""

# Import libraries 
import os
import pandas as pd
import glob

# -- PROVIDE PATH TO ICE DISPLACEMENT DATA (LEV5, LEV6, KAN_U, F003 & F004) --
data_path = '.../data' # UPDATE FILEPATH
folders = ['seasonal_annual_2009_2013', 'seasonal_annual_2020_2024']
```

```python
# CALCULATE AND PRINT STATISTICS REGARDING CHANGES IN ANNUAL ICE DISPLACEMENT BETWEEN 2009-2012 AND 2021-2023 (LEV5)
site = 'lev5' 
all_data = pd.concat([pd.read_csv(fs) for folder in folders for fs in glob.glob(os.path.join(data_path, folder, f'{site}_seasonal_annual_*.csv'))], axis=0)
data_2009_2012 = all_data[(all_data['year'].between(2009, 2012)) & (all_data['site'] == site) & (all_data['period'] == 'Annual')]
data_2021_2023 = all_data[(all_data['year'].between(2021, 2023)) & (all_data['site'] == site) & (all_data['period'] == 'Annual')]
mean_disp_2009_2012 = data_2009_2012['disp_m'].mean()
std_disp_2009_2012 = data_2009_2012['disp_m'].std()
mean_disp_2021_2023 = data_2021_2023['disp_m'].mean()
std_disp_2021_2023 = data_2021_2023['disp_m'].std()

midpoint_2009_2012 = 2010.5
midpoint_2021_2023 = 2022
time_difference = midpoint_2021_2023 - midpoint_2009_2012
percentage_change = ((mean_disp_2021_2023 - mean_disp_2009_2012) / mean_disp_2009_2012) * 100
percentage_change_rate_per_year = percentage_change / time_difference  

print(f"Site: {site.upper()}")
print(f"Mean Displacement (2009-2012): {mean_disp_2009_2012:.2f} m")
print(f"Standard Deviation (2009-2012): {std_disp_2009_2012:.2f} m")
print(f"Mean Displacement (2021-2023): {mean_disp_2021_2023:.2f} m")
print(f"Standard Deviation (2021-2023): {std_disp_2021_2023:.2f} m")
print(f"Total Percentage Change (2009-2012 to 2021-2023 mean): {percentage_change:.2f}%")
print(f"Percentage change per year (2010.5 to 2022): {percentage_change_rate_per_year:.2f}% per year\n")
```

```python
# CALCULATE AND PRINT STATISTICS REGARDING CHANGES IN ANNUAL ICE DISPLACEMENT BETWEEN 2009-2012 AND 2021-2023 (LEV6 & KAN_U)
for site in ['lev6', 'kanu']:
    all_data = pd.concat([pd.read_csv(fs) for folder in folders for fs in glob.glob(os.path.join(data_path, folder, f'{site}_seasonal_annual_*.csv'))], axis=0)
    data_2009_2012 = all_data[(all_data['year'].between(2009, 2012)) & (all_data['site'] == site) & (all_data['period'] == 'Annual')]
    data_2021_2023 = all_data[(all_data['year'].between(2021, 2023)) & (all_data['site'] == site) & (all_data['period'] == 'Annual')]
    mean_disp_2009_2012 = data_2009_2012['disp_m'].mean()
    std_disp_2009_2012 = data_2009_2012['disp_m'].std()
    mean_disp_2021_2023 = data_2021_2023['disp_m'].mean()
    std_disp_2021_2023 = data_2021_2023['disp_m'].std()

    midpoint_2009_2012 = 2010.5
    midpoint_2021_2023 = 2022
    time_difference = midpoint_2021_2023 - midpoint_2009_2012
    percentage_change = ((mean_disp_2021_2023 - mean_disp_2009_2012) / mean_disp_2009_2012) * 100
    percentage_change_rate_per_year = percentage_change / time_difference

    print(f"Site: {site.upper()}")
    print(f"Mean Displacement (2009-2012): {mean_disp_2009_2012:.2f} m")
    print(f"Standard Deviation (2009-2012): {std_disp_2009_2012:.2f} m")
    print(f"Mean Displacement (2021-2023): {mean_disp_2021_2023:.2f} m")
    print(f"Standard Deviation (2021-2023): {std_disp_2021_2023:.2f} m")
    print(f"Total Percentage Change (2009-2012 to 2021-2023 mean): {percentage_change:.2f}%")
    print(f"Percentage change per year (2010.5 to 2022): {percentage_change_rate_per_year:.2f}% per year\n")
```

```python
# Mean winter ice velocity, 2009-2012 vs 2021-2023 (LEV6 & KAN_U)
for site in ['lev6', 'kanu']:
    all_data = pd.concat([pd.read_csv(fs) for folder in folders for fs in glob.glob(os.path.join(data_path, folder, f'{site}_seasonal_annual_*.csv'))], axis=0)
    winter = all_data[(all_data['site'] == site) & (all_data['period'] == 'Winter')]
    w1 = winter[winter['year'].between(2009, 2012)]['vel_m_yr']
    w2 = winter[winter['year'].between(2021, 2023)]['vel_m_yr']
    print(f"{site.upper()}: 2009-2012 winters {w1.round(2).tolist()} mean {w1.mean():.2f} | "
          f"2021-2023 winters {w2.round(2).tolist()} mean {w2.mean():.2f} | change {100*(w2.mean()/w1.mean()-1):.2f}%")
```

```python
# CALCULATE AND PRINT STATISTICS REGARDING CHANGES IN ICE DISPLACEMENT BETWEEN 1996 AND 2021-2023
sites = ['f004', 'f003'] 
folder = ['seasonal_annual_2020_2024']
displacement_1996_values = {'f004': 54.0, 'f003': 41.1}  # 1996 PARCA annual displacements (not in the HI-SLIDE files)
for site in sites:
    all_data = pd.concat([pd.read_csv(fs) for folder in folder for fs in glob.glob(os.path.join(data_path, folder, f'{site}_seasonal_annual_*.csv'))], axis=0)
    # Add 1996 annual displacement if not already in the files
    if not ((all_data['year'] == 1996) & (all_data['period'] == 'Annual')).any():
        all_data = pd.concat([all_data, pd.DataFrame([{'site': site, 'period': 'Annual', 'year': 1996,
                              'disp_m': displacement_1996_values[site]}])], ignore_index=True)
    filtered_data = all_data[(all_data['year'].between(2021, 2023)) & (all_data['site'] == site) & (all_data['period'] == 'Annual')]
    mean_disp = filtered_data['disp_m'].mean() # Calculate mean displacement
    std_disp = filtered_data['disp_m'].std() # Calculate standard deviation
    displacement_1996 = all_data[(all_data['year'] == 1996) & (all_data['site'] == site) & (all_data['period'] == 'Annual')]['disp_m'].values
    if len(displacement_1996) > 0:
        displacement_1996 = displacement_1996[0] # Get displacement for 1996
        percentage_change = ((mean_disp - displacement_1996) / displacement_1996) * 100 # Calculate percentage change from 1996 to 2021-2023
        percentage_change_rate_per_year = percentage_change / (2022 - 1996)  # Calculate rate of percentage change per year (1996 - 2022)
        print(f"Site: {site.upper()}")
        print(f"Mean Displacement (2021-2023): {mean_disp:.2f} m")
        print(f"Standard Deviation (2021-2023): {std_disp:.2f} m")
        print(f"Ice Displacement in 1996: {displacement_1996:.2f} m")
        print(f"Total Percentage Change (1996 to 2021-2023 mean): {percentage_change:.2f}%")
        print(f"Percentage change per year (1996-2022): {percentage_change_rate_per_year:.2f}% per year\n")
    else:
        print(f"{site.upper()}: no 1996 value found - check input files\n")
```
